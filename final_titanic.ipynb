### Insight: Age Distribution (Histogram)

۱) توزیع سنی مسافران نشان می‌دهد که بیشترین فراوانی مربوط به افراد جوان، به‌ویژه در بازه‌ی ۲۰ تا ۳۰ سال است؛ این گروه بخش قابل‌توجهی از مسافران را تشکیل می‌دادند.

۲) تعداد کودکان (به‌خصوص زیر ۱۰ سال) کمتر است، اما با توجه به سیاست «زنان و کودکان اول»، بررسی ارتباط سن با احتمال نجات در مراحل بعدی اهمیت زیادی دارد.

۳) توزیع سن کاملاً نرمال نیست؛ پس از حدود ۵۰ سالگی فراوانی مسافران کاهش می‌یابد و نشان می‌دهد که افراد سالمند حضور کمتری در کشتی داشته‌اند.

۴) در انتهای توزیع، چند مقدار بسیار بالا دیده می‌شود که احتمالاً Outlier هستند؛ برای بررسی دقیق‌تر، استفاده از Boxplot مناسب‌تر است.

۵) منحنی KDE نشان می‌دهد که توزیع سن دارای یک قله‌ی اصلی (Unimodal) است و دامنه‌ی سنی مسافران گسترده بوده؛ این موضوع تنوع سنی مسافران را تأیید می‌کند.

</div>

```python
plt.figure(figsize=(8,4))
sns.histplot(data=df, x="Fare", bins=40, kde=True, color="teal")
plt.title("Histogram of Fare")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

<div dir="rtl">

### Insight: Fare Distribution (Histogram)

۱) توزیع قیمت بلیت (Fare) نشان می‌دهد که بخش زیادی از مسافران بلیت‌های ارزان‌قیمت داشته‌اند و بیشترین فراوانی در بازه‌های پایین قیمت مشاهده می‌شود.

۲) این الگو با تعداد بیشتر مسافران کلاس سوم (Pclass = 3) سازگار است؛ با این حال، برای بررسی دقیق رابطه‌ی بین Fare و Pclass باید تحلیل مشترک این دو متغیر انجام شود.

۳) توزیع Fare نرمال نیست و دارای چولگی مثبت (Right-Skewed) است؛ به این معنا که تعداد کمی از مسافران بلیت‌های بسیار گران داشته‌اند و دم توزیع به سمت راست کشیده شده است.

۴) در انتهای توزیع، تعداد کمی مقدار با قیمت‌های بسیار بالا مشاهده می‌شود. برای بررسی اینکه آیا این مقادیر Outlier محسوب می‌شوند یا خیر، استفاده از Boxplot مناسب‌تر است.

۵) منحنی KDE نشان می‌دهد که توزیع Fare دارای یک قله‌ی اصلی (Unimodal) است و دامنه‌ی قیمت بلیت‌ها نسبتاً گسترده است؛ این موضوع بیانگر تنوع قابل‌توجه قیمت بلیت‌ها در میان مسافران است.

</div>

```python
fig = plt.figure(figsize=(12,8))

# Age
ax1 = fig.add_subplot(2, 2, 1)
sns.boxplot(data=df, y="Age", color="orchid", ax=ax1)
ax1.set_title("Age")

# Fare
ax2 = fig.add_subplot(2, 2, 2)
sns.boxplot(data=df, y="Fare", color="teal", ax=ax2)
ax2.set_title("Fare")

# Siblings/Spouses Aboard
ax3 = fig.add_subplot(2, 2, 3)
sns.boxplot(data=df, y="Siblings/Spouses Aboard", color="gold", ax=ax3)
ax3.set_title("Siblings/Spouses Aboard")

# Parents/Children Aboard
ax4 = fig.add_subplot(2, 2, 4)
sns.boxplot(data=df, y="Parents/Children Aboard", color="salmon", ax=ax4)
ax4.set_title("Parents/Children Aboard")

plt.tight_layout()
plt.show()
```

<div dir="rtl">

### Insight: Boxplot Analysis of Numerical Features

۱) Age  
نمودار Boxplot مربوط به Age نشان می‌دهد که بیشتر مسافران در بازه‌ی سنی جوان تا میانسال قرار دارند. همچنین چند مقدار در سنین بالاتر به‌صورت Outlier مشاهده می‌شود که بیانگر حضور تعداد کمی از مسافران سالمند در این دیتاست است.

۲) Fare  
نمودار Fare وجود چند Outlier با قیمت‌های بسیار بالا را نشان می‌دهد. همچنین پراکندگی زیاد داده‌ها و فاصله‌ی قابل‌توجه بین میانه و مقادیر بزرگ‌تر، بیانگر چولگی مثبت (Right-Skewed) در توزیع قیمت بلیت‌ها است.

۳) Siblings/Spouses Aboard  
بیشتر داده‌ها در مقدار ۰ متمرکز هستند که نشان می‌دهد اکثر مسافران بدون خواهر، برادر یا همسر سفر کرده‌اند. پراکندگی این متغیر نسبتاً کم است و تنها تعداد محدودی از مسافران با یک یا چند همراه در کشتی حضور داشته‌اند.

۴) Parents/Children Aboard  
در این ستون نیز بیشترین فراوانی مربوط به مقدار ۰ است و تنها تعداد کمی از مسافران همراه با والدین یا فرزندان سفر کرده‌اند. این موضوع نشان می‌دهد که بیشتر مسافران به‌تنهایی یا بدون والدین و فرزندان در سفر بوده‌اند.

### جمع‌بندی

چهار نمودار Boxplot نشان می‌دهند که ویژگی‌های عددی دیتاست Titanic الگوهای پراکندگی متفاوتی دارند:

- Age و Fare دارای پراکندگی بیشتر بوده و چند Outlier قابل‌مشاهده دارند.  
- Siblings/Spouses Aboard و Parents/Children Aboard عمدتاً در مقادیر پایین، به‌ویژه مقدار ۰، متمرکز هستند و پراکندگی کمتری دارند.

این ویژگی‌ها می‌توانند در مراحل بعدی تحلیل داده، بررسی Outlier و مدل‌سازی نقش مهمی داشته باشند.

</div>

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

### Insight: Correlation Heatmap – Titanic

۱) Survived و Pclass  
همبستگی منفی نسبتاً قوی بین Survived و Pclass (حدود -0.34) نشان می‌دهد که با افزایش مقدار Pclass (از کلاس اول به کلاس سوم)، احتمال نجات کاهش می‌یابد. این متغیر یکی از مهم‌ترین ویژگی‌های عددی برای تحلیل نجات در این دیتاست است.

۲) Pclass و Age  
بین Pclass و Age همبستگی منفی قابل‌توجهی (حدود -0.39) مشاهده می‌شود؛ به این معنا که مسافران کلاس‌های بالاتر به‌طور میانگین سن بیشتری داشته‌اند. این رابطه می‌تواند در تحلیل‌های بعدی مورد بررسی دقیق‌تر قرار گیرد.

۳) Age و Survived  
همبستگی بین Age و Survived بسیار ضعیف و نزدیک به صفر (حدود -0.06) است. این موضوع نشان می‌دهد که در سطح کل دیتاست، رابطه‌ی خطی قابل‌توجهی بین سن و احتمال نجات مشاهده نمی‌شود.

۴) Siblings Aboard و سایر متغیرها  
همبستگی Siblings Aboard با Survived بسیار ضعیف (حدود -0.04) است و نشان می‌دهد رابطه‌ی خطی محسوسی بین این دو متغیر وجود ندارد. همچنین همبستگی منفی آن با Age بیانگر این است که افراد مسن‌تر، به‌طور میانگین، همراهان کمتری داشته‌اند.

### جمع‌بندی

Heatmap نشان می‌دهد که قوی‌ترین روابط خطی در میان متغیرهای عددی این دیتاست مربوط به:

- Pclass ↔ Survived
- Pclass ↔ Age

در مقابل، متغیرهایی مانند Age و Siblings Aboard همبستگی خطی ضعیفی با Survived دارند. بنابراین، Pclass یکی از مهم‌ترین ویژگی‌های عددی برای تحلیل این دیتاست محسوب می‌شود. برای ارزیابی اهمیت سایر متغیرها، بررسی‌های تکمیلی مانند تحلیل‌های چندمتغیره یا مدل‌سازی می‌تواند مفید باشد.
